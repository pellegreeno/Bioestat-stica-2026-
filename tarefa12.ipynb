#Questão 2

a) A distribuição utilizada deve ser a binomial, pois são 10 capturas independentes, cada uma com dois resultados possíveis (com a característica genética ou não), e conta-se o número de sucessos.

  

```python
import math

def pmf_binomial(k, n, p):
    """Calculates the probability mass function for a binomial distribution."""
    return math.comb(n, k) * (p**k) * ((1-p)**(n-k))

print("\n--- Questão 2: Binomial(n=10, p=0.3) ---")
n, p = 10, 0.3
print(f"2b) P(X=3):            {pmf_binomial(3, n, p):.4f}")
print(f"2c) P(X>=1) = 1-P(0):  {1 - pmf_binomial(0, n, p):.4f}")
```

#Questão 3

a) Poisson, com λ = 2 encontros/hora. São contagens de eventos independentes em um intervalo de tempo, com taxa média constante.

```python
import math

def pmf_poisson(k, lam):
    """Calculates the probability mass function for a Poisson distribution."""
    return (lam**k * math.exp(-lam)) / math.factorial(k)

print("\n--- Questão 3: Poisson(lambda=2) ---")
lam = 2
print(f"3b) P(X=3):  {pmf_poisson(3, lam):.4f}")
print(f"3c) P(X=0):  {pmf_poisson(0, lam):.4f}")
p_ate_2 = sum(pmf_poisson(k, lam) for k in range(3))   # P(0)+P(1)+P(2)
print(f"3d) P(X>2) = 1 - P(X<=2): {1 - p_ate_2:.4f}")
```
