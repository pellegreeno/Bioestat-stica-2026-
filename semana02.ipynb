```python
import pandas as pd

# Carregue seu arquivo Excel aqui.
df = pd.read_excel('/content/dados_semana02 (1).xlsx') # Substitua 'seuarquivo.csv' pelo nome do seu arquivo

# Mostra as primeiras 5 linhas do DataFrame
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Contar a frequência de cada idade
age_counts = df['IDADE (ANOS)'].value_counts().sort_index()

# Criar o gráfico de pizza
fig = plt.figure(figsize=(10, 8))
plt.pie(age_counts, labels=age_counts.index, autopct='%1.1f%%', startangle=90, wedgeprops={'edgecolor': 'black'})
plt.title('Distribuição de Idades (Anos)')
plt.axis('equal')  # Garante que o círculo seja desenhado como um círculo.
plt.show()
```

```python
species_counts = df['ESPÉCIE'].value_counts()

fig = plt.figure(figsize=(12, 6))
sns.barplot(x=species_counts.index, y=species_counts.values, palette='viridis')
plt.title('Distribuição de Espécies')
plt.xlabel('Espécie')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(8, 6))
sns.boxplot(y=df['MASSA (kg)'], palette='coolwarm')
plt.title('Distribuição da Massa (kg)')
plt.ylabel('Massa (kg)')
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(10, 6))
sns.histplot(df['TEMP CORPORAL (°C)'], kde=True, color='skyblue')
plt.title('Distribuição da Temperatura Corporal (°C)')
plt.xlabel('Temperatura Corporal (°C)')
plt.ylabel('Frequência')
plt.tight_layout()
plt.show()
```
