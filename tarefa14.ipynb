# **Exercício 1**

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Parâmetros da distribuição
mu = 420
sigma = 55

# 1. Cálculos de Probabilidade
p1 = stats.norm.cdf(350, mu, sigma)  # P(X < 350)
p2 = 1 - stats.norm.cdf(500, mu, sigma)  # P(X > 500)
p3 = stats.norm.cdf(480, mu, sigma) - stats.norm.cdf(
    400, mu, sigma
)  # P(400 <= X <= 480)

print(f"1. P(X < 350 g) = {p1:.4f} ({p1*100:.2f}%)")
print(f"2. P(X > 500 g) = {p2:.4f} ({p2*100:.2f}%)")
print(f"3. P(400 <= X <= 480 g) = {p3:.4f} ({p3*100:.2f}%)")

# 4. Representação Gráfica
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

fig, axs = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

# Gráfico 1: X < 350
axs[0].plot(x, y, "black", lw=1.5)
axs[0].fill_between(x, y, where=(x <= 350), color="skyblue", alpha=0.6)
axs[0].set_title(f"P(X < 350) = {p1*100:.1f}%")
axs[0].set_xlabel("Massa Corporal (g)")
axs[0].set_ylabel("Densidade de Probabilidade")

# Gráfico 2: X > 500
axs[1].plot(x, y, "black", lw=1.5)
axs[1].fill_between(x, y, where=(x >= 500), color="salmon", alpha=0.6)
axs[1].set_title(f"P(X > 500) = {p2*100:.1f}%")
axs[1].set_xlabel("Massa Corporal (g)")

# Gráfico 3: 400 <= X <= 480
axs[2].plot(x, y, "black", lw=1.5)
axs[2].fill_between(
    x, y, where=((x >= 400) & (x <= 480)), color="lightgreen", alpha=0.6
)
axs[2].set_title(f"P(400 ≤ X ≤ 480) = {p3*100:.1f}%")
axs[2].set_xlabel("Massa Corporal (g)")

plt.tight_layout()
plt.show()
```

Resultados$P(X < 350\text{ g})$: $10{,}16\%$$P(X > 500\text{ g})$: $7{,}29\%$$P(400\text{ g} \le X \le 480\text{ g})$: $50{,}43\%$

# **Exercício 2**

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

mu = 150
sigma = 10.5

# 1 e 2. Cálculo dos percentis
p80 = stats.norm.ppf(0.80, mu, sigma)
p90 = stats.norm.ppf(
    0.90, mu, sigma
)  # 10% superior equivale ao percentil 90 (P90)
p50 = stats.norm.ppf(0.50, mu, sigma)  # Mediana / Média

print(f"Percentil 80 (P80): {p80:.2f} cm")
print(f"Limiar dos 10% maiores (P90): {p90:.2f} cm")

# 3. Gráfico Único
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

plt.figure(figsize=(9, 5))
plt.plot(x, y, "k-", lw=2, label="Distribuição Normal")

plt.axvline(
    p50,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"P50 (Média/Mediana): {p50:.1f} cm",
)
plt.axvline(
    p80, color="orange", linestyle="--", linewidth=2, label=f"P80: {p80:.2f} cm"
)
plt.axvline(
    p90, color="red", linestyle="--", linewidth=2, label=f"P90: {p90:.2f} cm"
)

plt.fill_between(
    x, y, where=(x >= p90), color="red", alpha=0.3, label="10% Maiores Indivíduos"
)

plt.title("Percentis de Comprimento Corporal dos Mamíferos")
plt.xlabel("Comprimento Corporal (cm)")
plt.ylabel("Densidade de Probabilidade")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

**Interpretação Biológica**

Percentil 80 ($158{,}84\text{ cm}$): $80\%$ dos indivíduos da população possuem comprimento corporal menor ou igual a $158{,}84\text{ cm}$, enquanto apenas $20\%$ superam esse tamanho.

Limiar dos $10\%$ maiores / Percentil 90 ($163{,}46\text{ cm}$): O valor de $163{,}46\text{ cm}$ é a fronteira que separa os $10\%$ maiores indivíduos dos $90\%$ restantes.

Significado Ecológico: Indivíduos acima do percentil 90 podem apresentar vantagens competitivas em disputas territoriais, seleção sexual ou predação, enquanto exigem maior suprimento calórico diário.

# **Exercício 3**

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

mu = 72
sigma = 8

# 1 a 3. Probabilidades
p_le_60 = stats.norm.cdf(60, mu, sigma)
p_ge_85 = 1 - stats.norm.cdf(85, mu, sigma)
p_65_80 = stats.norm.cdf(80, mu, sigma) - stats.norm.cdf(65, mu, sigma)

# 4. Percentis
p90 = stats.norm.ppf(0.90, mu, sigma)
p95 = stats.norm.ppf(0.95, mu, sigma)
p99 = stats.norm.ppf(0.99, mu, sigma)

print("--- RESULTADOS - EXERCÍCIO 3 ---")
print(f"1. P(X <= 60 g) = {p_le_60:.4f} ({p_le_60*100:.2f}%)")
print(f"2. P(X >= 85 g) = {p_ge_85:.4f} ({p_ge_85*100:.2f}%)")
print(f"3. P(65 <= X <= 80 g) = {p_65_80:.4f} ({p_65_80*100:.2f}%)")
print(
    f"4. Percentis:\n   - P90: {p90:.2f} g\n   - P95: {p95:.2f} g\n   - P99: {p99:.2f} g\n"
)

# 5. Gráfico Integrado
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Regiões de Probabilidade
ax1.plot(x, y, "k-", lw=1.5)
ax1.fill_between(
    x, y, where=(x <= 60), color="blue", alpha=0.4, label="P(X <= 60g) [6.7%]"
)
ax1.fill_between(
    x,
    y,
    where=((x >= 65) & (x <= 80)),
    color="green",
    alpha=0.4,
    label="P(65 <= X <= 80g) [65.1%]",
)
ax1.fill_between(
    x, y, where=(x >= 85), color="red", alpha=0.4, label="P(X >= 85g) [5.2%]"
)
ax1.set_title("Regiões de Probabilidade de Massa Corporal")
ax1.set_ylabel("Densidade")
ax1.legend()
ax1.grid(alpha=0.3)

# Percentis
ax2.plot(x, y, "k-", lw=1.5)
ax2.axvline(
    p90,
    color="orange",
    linestyle="--",
    lw=2,
    label=f"P90 ({p90:.1f}g) - Top 10%",
)
ax2.axvline(
    p95, color="magenta", linestyle="--", lw=2, label=f"P95 ({p95:.1f}g) - Top 5%"
)
ax2.axvline(
    p99, color="purple", linestyle="--", lw=2, label=f"P99 ({p99:.1f}g) - Top 1%"
)
ax2.set_title("Percentis Extremos (P90, P95, P99)")
ax2.set_xlabel("Massa Corporal (g)")
ax2.set_ylabel("Densidade")
ax2.legend()
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Resultados Obtidos $P(X \le 60\text{ g}) = 0{,}0668$ ($6{,}68\%$)$P(X \ge 85\text{ g}) = 0{,}0521$ ($5{,}21\%$)$P(65\text{ g} \le X \le 80\text{ g}) = 0{,}6506$ ($65{,}06\%$)

Percentis: $P_{90} = 82{,}25\text{ g}$, $P_{95} = 85{,}16\text{ g}$, $P_{99} = 90{,}61\text{ g}$

**Interpretação Ecológica Corta**: A maior parte da população ($65{,}06\%$) concentra-se na faixa intermediária de massa ($65\text{ g}$ a $80\text{ g}$), caracterizando o padrão modal típico da espécie sob condições ambientais normais.
Aves abaixo de $60\text{ g}$ ($6{,}68\%$) representam indivíduos com possível desnutrição, estresse energético ou juvenis, sofrendo maior pressão de seleção por menor sobrevivência no inverno. Aves extremamente pesadas ($> 90{,}61\text{ g}$, $P_{99}$) são raras ($1\%$) e podem enfrentar uma compensação ecológica (trade-off): maior resiliência à escassez de alimento versus menor manobrabilidade de voo e maior vulnerabilidade à predação.
