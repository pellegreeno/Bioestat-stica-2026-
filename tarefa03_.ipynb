```python
import pandas as pd

# Load the Excel file
df = pd.read_excel('/content/tarefa_03.xlsx')

# Display the first 5 rows of the DataFrame
display(df)

# Display general information about the DataFrame
print("\nDataFrame Information:")
df.info()
```

```python
frequencia_simples_grupo_taxonomico = df['GRUPO TAXONÔMICO'].value_counts()

print("Frequência simples para 'GRUPO TAXONÔMICO':")
display(frequencia_simples_grupo_taxonomico)
```

```python
frequencia_simples_sexo = df['SEXO'].value_counts()

print("Frequência simples para 'SEXO':")
display(frequencia_simples_sexo)
```

```python
frequencia_relativa_grupo_taxonomico = df['GRUPO TAXONÔMICO'].value_counts(normalize=True) * 100

print("Frequência relativa para 'GRUPO TAXONÔMICO' (%):")
display(frequencia_relativa_grupo_taxonomico)
```

```python
frequencia_relativa_sexo = df['SEXO'].value_counts(normalize=True) * 100

print("Frequência relativa para 'SEXO' (%):")
display(frequencia_relativa_sexo)
```

### Interpretação das Tabelas de Frequência

**1. Frequência de 'GRUPO TAXONÔMICO':**

**Interpretação:** Com base nas frequências simples, 'Mamíferos' são o grupo mais observado com 7 ocorrências, seguido por 'Aves' e 'Insetos', ambos com 6 ocorrências. Em termos de frequência relativa, 'Mamíferos' representam aproximadamente 36.84% do total, enquanto 'Aves' e 'Insetos' contribuem com cerca de 31.58% cada. Isso sugere uma distribuição relativamente equilibrada entre os grupos, com uma ligeira predominância de mamíferos na amostra.

**2. Frequência de 'SEXO':**

**Interpretação:** A frequência simples mostra 11 indivíduos do sexo 'M' e 8 do sexo 'F'. Relativamente, o sexo 'M' representa aproximadamente 57.89% das observações, enquanto o sexo 'F' representa cerca de 42.11%. Há uma predominância de indivíduos do sexo masculino no conjunto de dados, o que pode indicar um viés na amostragem ou diferenças populacionais.

**3. Frequência de 'FASE DE DESENVOLVIMENTO':**

**Interpretação:** A fase adulta é a mais frequente, com 8 indivíduos (aproximadamente 42.11%). A fase de 'filhote' segue com 6 indivíduos (cerca de 31.58%), e a fase 'jovem' tem 5 indivíduos (aproximadamente 26.32%). Isso é um padrão comum em populações biológicas, onde adultos podem ser mais numerosos ou mais facilmente detectáveis do que jovens ou filhotes.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gráfico de Barras para 'GRUPO TAXONÔMICO'
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='GRUPO TAXONÔMICO', order=df['GRUPO TAXONÔMICO'].value_counts().index)
plt.title("Gráfico de Barras: Frequência de 'GRUPO TAXONÔMICO'")
plt.xlabel('Grupo Taxonômico')
plt.ylabel('Contagem')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Histogramas para Variáveis Numéricas

Vamos agora gerar histogramas para visualizar a distribuição das variáveis numéricas 'TAMANHO (cm)' e 'PESO (g)'.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histograma para 'TAMANHO (cm)'
plt.figure(figsize=(10, 6))
sns.histplot(df['TAMANHO (cm)'], kde=True, bins=5, color='skyblue')
plt.title("Histograma: Distribuição de 'TAMANHO (cm)'")
plt.xlabel('Tamanho (cm)')
plt.ylabel('Frequência')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gráfico de Dispersão para 'TAMANHO (cm)' vs 'PESO (g)'
plt.figure(figsize=(10, 7))
sns.scatterplot(data=df, x='TAMANHO (cm)', y='PESO (g)', hue='GRUPO TAXONÔMICO', palette='viridis', s=100)
plt.title("Gráfico de Dispersão: 'PESO (g)' vs 'TAMANHO (cm)'")
plt.xlabel('Tamanho (cm)')
plt.ylabel('Peso (g)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Grupo Taxonômico')
plt.tight_layout()
plt.show()
```
