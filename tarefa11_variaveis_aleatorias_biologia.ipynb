# Tarefa 11: Variáveis Aleatórias e Modelos Probabilísticos Aplicados à Biologia


```python
# Importação das bibliotecas essenciais
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configuração para reprodutibilidade das simulações
np.random.seed(42)

# Configuração estética dos gráficos
sns.set_theme(style="whitegrid")
```

## 1. Variável Aleatória

Uma variável aleatória (\(X\)) é uma função matemática que atribui um valor numérico ao resultado de um experimento aleatório. Em Biologia, os experimentos envolvem incertezas inerentes aos sistemas vivos, como variações genéticas, ambientais ou demográficas.

### Exemplo Aplicado à Biologia:
* **Experimento aleatório:** Amostragem de sementes coletadas de uma árvore da Mata Atlântica para teste de viabilidade.
* **Variável aleatória (\(X\)):** Número de sementes que germinam em um lote de \(n\) sementes.
* **Valores possíveis:** são X ∈ {0, 1, 2,...,n}

## 2. Variáveis Aleatórias Discretas e Contínuas

* **Variáveis Discretas:** Assumem valores contáveis (geralmente inteiros não negativos). Resultam de processos de contagem.
* **Variáveis Contínuas:** Assumem qualquer valor dentro de um intervalo real. Resultam de processos de medição.

### Exemplos Contextualizados:

| Tipo de Variável | Exemplos em Biologia | Justificativa |
| :--- | :--- | :--- |
| **Discreta** | 1. Número de indivíduos de uma espécie em uma parcela.

## 3. Modelo de Bernoulli

O modelo de Bernoulli descreve um experimento aleatório com exatamente dois resultados possíveis:
* **Sucesso (\(X = 1\)):** O evento de interesse ocorre (com probabilidade \(p\)).
* **Fracasso (\(X = 0\)):** O evento não ocorre (com probabilidade \(1 - p\)).

### Aplicação Biológica:
Avaliar se uma bactéria isolada de um ambiente hospitalar apresenta ou não resistência a um determinado antibiótico (\(p = 0.30\) ou 30% de chance de resistência na população estudada).

```python
# Parâmetro de probabilidade de sucesso (resistência bacteriana)
p_sucesso = 0.30

# Simulação de 10 testes bacterianos individuais (0 = sensível, 1 = resistente)
amostra_bernoulli = np.random.binomial(n=1, p=p_sucesso, size=10)

print("Resultados da amostra de bactérias (0 = Sensível, 1 = Resistente):")
print(amostra_bernoulli)

# Interpretação biológica
resistentes = np.sum(amostra_bernoulli)
print(f"\nTotal de bactérias resistentes encontradas na amostra: {resistentes} de 10.")
```

## 4. Modelo Binomial

O modelo Binomial estende o conceito de Bernoulli para contar o número total de sucessos (X) em \(n\) ensaios independentes, mantendo a probabilidade de sucesso \(p\) constante.

### Aplicação Biológica:
Determinar a probabilidade de germinação de sementes. Se plantarmos \(n = 20\) sementes com uma taxa de germinação esperada de \(p = 0.75\), qual é a distribuição de probabilidade do número de sementes germinadas?

```python
# Parâmetros do modelo Binomial
n_sementes = 20
p_germinacao = 0.75

# Geração de valores possíveis para a variável aleatória X
k_valores = np.arange(0, n_sementes + 1)
probabilidades = stats.binom.pmf(k_valores, n_sementes, p_germinacao)

# Visualização gráfica da distribuição binomial
plt.figure(figsize=(10, 5))
plt.bar(k_valores, probabilidades, color='forestgreen', alpha=0.8, edgecolor='black')
plt.title('Distribuição Binomial: Probabilidade de Germinação de Sementes (n=20, p=0.75)', fontsize=12)
plt.xlabel('Número de Sementes Germinadas (\(X\))', fontsize=10)
plt.ylabel('Probabilidade P(X = k)', fontsize=10)
plt.xticks(k_valores)
plt.show()

# Cálculo do valor esperado (média) e desvio padrão
media_teo = n_sementes * p_germinacao
desvio_teo = np.sqrt(n_sementes * p_germinacao * (1 - p_germinacao))
print(f"Número médio esperado de sementes germinadas: {media_teo:.2f}")
print(f"Desvio padrão: {desvio_teo:.2f}")
```

## 5. Modelo de Poisson

O modelo de Poisson modela a contagem de eventos raros ou discretos em um intervalo contínuo de tempo, espaço ou área, definido por uma taxa média de ocorrência λ.

### Aplicação Biológica:
Contagem do número de indivíduos de uma espécie de ave observados por parcela de inventário florestal de (1 km²), sabendo que a densidade média esperada é de (λ = 3.5\) aves por parcela.

```python
# Parâmetro da taxa média (lambda)
lambda_taxa = 3.5

# Geração de valores de contagem
k_poisson = np.arange(0, 12)
prob_poisson = stats.poisson.pmf(k_poisson, lambda_taxa)

# Gráfico da Distribuição de Poisson
plt.figure(figsize=(10, 5))
plt.plot(k_poisson, prob_poisson, marker='o', color='darkorange', linewidth=2, markersize=8)
plt.bar(k_poisson, prob_poisson, color='darkorange', alpha=0.3, edgecolor='black')
plt.title('Distribuição de Poisson: Número de Aves por Parcela de Amostragem (\(\lambda = 3.5\))', fontsize=12)
plt.xlabel('Número de Indivíduos Observados (\(X\))', fontsize=10)
plt.ylabel('Probabilidade P(X = k)', fontsize=10)
plt.xticks(k_poisson)
plt.show()
```
