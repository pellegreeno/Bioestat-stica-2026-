```python
import pandas as pd

# Carregue seu arquivo Excel aqui.
df = pd.read_excel('/content/dados_semana02 .xlsx') # Substitua 'seuarquivo.csv' pelo nome do seu arquivo

# Mostra as primeiras 5 linhas do DataFrame
df
```

```python
import matplotlib.pyplot as plt

# Get the value counts of the 'IDADE (ANOS)' column
especie_counts = df['ESPÉCIE'].value_counts()

# Create the pie chart
plt.figure(figsize=(10, 8))
plt.pie(especie_counts, labels=especie_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribuição das Espécies')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Get the value counts of the 'IDADE (ANOS)' column
idade_counts = df['IDADE (ANOS)'].value_counts().reset_index()
idade_counts.columns = ['IDADE (ANOS)', 'Count']

# Create the bar chart
plt.figure(figsize=(8, 6))
sns.barplot(x='IDADE (ANOS)', y='Count', data=idade_counts, palette='viridis')
plt.title('Distribuição de Idades (Anos)')
plt.xlabel('Idade (Anos)')
plt.ylabel('Contagem')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create the boxplot for 'MASSA (kg)'
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['MASSA (kg)'], palette='coolwarm')
plt.title('Distribuição da Massa (kg)')
plt.ylabel('Massa (kg)')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create the histogram for 'TEMP CORPORAL (°C)'
plt.figure(figsize=(10, 6))
sns.histplot(df['TEMP CORPORAL (°C)'], kde=True, bins=10, color='skyblue')
plt.title('Distribuição da Temperatura Corporal (°C)')
plt.xlabel('Temperatura Corporal (°C)')
plt.ylabel('Frequência')
plt.show()
```
